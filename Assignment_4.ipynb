```python
import pandas as pd
import numpy as np
df=pd.read_csv("Property Sales of Melbourne City.csv")
print(df.head())
print(df.describe())
print(df.tail())
print(df.info())
```

```python
range_a = df['Price'].max() - df['Price'].min() 
range_b = df['Landsize'].max() - df['Landsize'].min() 
print('Range - Price:', range_a) 
print('Range - Landsize:', range_b) 
```

```python
# 3. Variance 
var_a = df['Price'].var() 
var_b = df['Landsize'].var() 

print('Variance - Price:', var_a) 
print('Variance - Landsize:', var_b) 
```

```python
# 4. Standard Deviation 
std_a = df['Price'].std() 
std_b = df['Landsize'].std() 

print('Std Dev - Price:', std_a) 
print('Std Dev - Landsize:', std_b) 
```

```python
# 5. Coefficient of Variation
cv_a = (std_a / df['Price'].mean()) * 100 
cv_b = (std_b / df['Landsize'].mean()) * 100 

print('CV% - Price:', cv_a) 
print('CV% - Landsize:', cv_b) 
```

```python
# 6. Interpretation 
if std_a < std_b: 
    print('Price has more consistent values than Landsize') 
else: 
    print('Landsize has more consistent values than Price')
```

```python
import pandas as pd
import numpy as np

A = [10, 12, 15, 18, 20]
B = [8, 10, 12, 14, 16]
```

```python
range_a = max(A) - min(A)
range_b = max(B) - min(B)

print('Range - A:', range_a)
print('Range - B:', range_b)
```

```python
var_a = np.var(A)
var_b = np.var(B)

print('Variance - A:', var_a)
print('Variance - B:', var_b)
```

```python
std_a = np.std(A)
std_b = np.std(B)

print('Std Dev - A:', std_a)
print('Std Dev - B:', std_b)
```

```python
cv_a = (std_a / np.mean(A)) * 100
cv_b = (std_b / np.mean(B)) * 100

print('CV% - A:', cv_a)
print('CV% - B:', cv_b)
```

```python
if std_a < std_b:
    print('A has more consistent values than B')
else:
    print('B has more consistent values than A')
```
