```python
import numpy as np
from scipy import stats
marks = [45, 60, 55, 70, 60, 80, 50, 60, 75, 65]
Mean=np.mean(marks)
Median=np.median(marks)
Mode=stats.mode(marks)[0]
print("Marks:",marks)
print("Mean:",Mean)
print("Median:",Median)
print("Mode:",Mode)
```

```python
import pandas as pd
import numpy as np
from scipy import stats
df=pd.read_csv(r"C:\Users\Admin\Downloads\Covid 19 in india.csv")
print(df.head())
```

```python
print(df.describe())
```

```python
# mean,median,mode using pandas
mean_val=df["Positive"].mean()
median_val=df["Positive"].median()
mode_val=df["Positive"].mode()[0]
print("Mean:",mean_val)
print("Median:",median_val)
print("Mode:",mode_val)
```

```python
# using numpy/scipy
positive_array=df["Positive"].dropna().to_numpy()
print("Numpy Mean:",np.mean(positive_array))
print("Numpy Median:",np.median(positive_array))
print('SciPy Mode:', stats.mode(positive_array).mode)
```

```python
#Comparing measures to understand data distribution/skewness 
print('Skewness:', df['Positive'].skew()) 
if mean_val > median_val: 
    print('Distribution is right (positively) skewed') 
elif mean_val < median_val: 
    print('Distribution is left (negatively) skewed') 
else: 
    print('Distribution is approximately symmetric')
```

```python
# Central tendency grouped by State

group = df.groupby('State')['Positive']

print("Mean:")
print(group.mean())

print("\nMedian:")
print(group.median())

print("\nMode:")
print(group.apply(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan))
```
