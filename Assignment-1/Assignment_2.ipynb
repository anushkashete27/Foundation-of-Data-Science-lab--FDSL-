```python
import pandas as pd
import numpy as np

# Sample sales dataset 
data = {
    "OrderID": range(1, 11),
    "Customer": ["Alice", "Bob", "Charlie", "Alice", "Dave", "Bob", "Eve", "Charlie", "Alice", "Dave"],
    "Region": ["East", "West", "East", "East", "South", "West", "North", "East", "West", "South"],
    "Product": ["Widget", "Gadget", "Widget", "Gizmo", "Widget", "Gadget", "Gizmo", "Gadget", "Widget", "Gizmo"],
    "Quantity": [10, 5, 8, 3, np.nan, 7, 12, 4, 6, 9],
    "Price": [2.5, 15.0, 2.5, 7.0, 2.5, 15.0, 7.0, 15.0, 2.5, 7.0],
}
df = pd.DataFrame(data)
df
```

```python
print("First 3 rows:")
display(df.head(3))

print("\nLast 3 rows:")
display(df.tail(3))

print("\nInfo:")
df.info()

print("\nDescribe:")
display(df.describe())

print("\nShape (rows, columns):", df.shape)
```

```python
# loc: label-based selection
print("loc example (rows 0-2, specific columns):")
display(df.loc[0:2, ["Customer", "Product", "Price"]])

# iloc: position-based selection
print("\niloc example (first 3 rows, first 3 columns):")
display(df.iloc[0:3, 0:3])

# Boolean indexing: filter rows where Region is 'East'
print("\nBoolean indexing (Region == 'East'):")
display(df[df["Region"] == "East"])
```

```python
print("Sum of Quantity by Region:")
display(df.groupby("Region")["Quantity"].sum())

print("\nMean Price by Product:")
display(df.groupby("Product")["Price"].mean())

print("\nMultiple aggregations with agg():")
display(df.groupby("Region").agg(
    total_qty=("Quantity", "sum"),
    avg_price=("Price", "mean"),
    order_count=("OrderID", "count")
))
```

```python
pivot = pd.pivot_table(
    df,
    values="Quantity",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)
pivot
```

```python
print("Missing values per column:")
display(df.isnull().sum())

print("\nRows with any missing values:")
display(df[df.isnull().any(axis=1)])

print("\nDropping rows with missing values (dropna):")
display(df.dropna())

print("\nFilling missing Quantity with the column mean (fillna):")
df_filled = df.copy()
df_filled["Quantity"] = df_filled["Quantity"].fillna(df_filled["Quantity"].mean())
display(df_filled)
```

```python
df_t = df_filled.copy()

# apply(): create a new column using a custom function
df_t["Total"] = df_t.apply(lambda row: row["Quantity"] * row["Price"], axis=1)

# map(): map values in a column to new values
region_codes = {"East": "E", "West": "W", "South": "S", "North": "N"}
df_t["RegionCode"] = df_t["Region"].map(region_codes)

# astype(): change a column's data type
df_t["Quantity"] = df_t["Quantity"].astype(int)

# rename(): rename columns
df_t = df_t.rename(columns={"Total": "TotalRevenue"})

# sort_values(): sort rows by a column
df_t = df_t.sort_values(by="TotalRevenue", ascending=False)

df_t
```

```python
# A second DataFrame with extra customer info, to merge with df
customer_info = pd.DataFrame({
    "Customer": ["Alice", "Bob", "Charlie", "Dave", "Eve"],
    "City": ["New York", "Chicago", "Boston", "Houston", "Miami"],
    "MembershipLevel": ["Gold", "Silver", "Gold", "Bronze", "Silver"]
})

merged_df = pd.merge(df, customer_info, on="Customer", how="left")
print("merge() result:")
display(merged_df.head())

joined_df = df.set_index("Customer").join(customer_info.set_index("Customer"), how="left")
print("\njoin() result:")
display(joined_df.head())

batch1 = df.iloc[:5]
batch2 = df.iloc[5:]
concatenated_df = pd.concat([batch1, batch2], ignore_index=True)
print("\nconcat() result (recombined batches):")
display(concatenated_df)
```
