```python
NumPy Theory – Simple Language for Practical Exam
1. NumPy

NumPy is a Python library used for numerical calculations. It helps us work easily with numbers and arrays.

2. Role in Scientific Computing

NumPy is the base library for many Python libraries like Pandas, SciPy, and scikit-learn. It provides fast and memory-efficient calculations.

3. ndarray

ndarray is the main data structure in NumPy. It is a multidimensional array where all elements usually have the same data type.

4. Array Creation

We can create arrays using:

np.array() – creates an array
np.zeros() – creates an array with zeros
np.ones() – creates an array with ones
np.arange() – creates numbers in a range
np.linspace() – creates equally spaced values
np.eye() – creates an identity matrix
5. Array Attributes

Important array attributes are:

shape – tells rows and columns
size – tells total number of elements
ndim – tells number of dimensions
dtype – tells the data type of elements
6. Indexing and Slicing

Indexing is used to access a particular element of an array.

Slicing is used to access a part of an array.

7. Reshaping and Stacking
reshape() – changes the shape of an array
flatten() – converts an array into one dimension
hstack() – joins arrays horizontally
vstack() – joins arrays vertically
8. Vectorized Operations and Broadcasting

NumPy allows us to perform calculations on the whole array at once without using loops.

Broadcasting allows operations between arrays of different shapes when possible.

9. Universal Functions (ufuncs)

Universal functions perform operations on each element of an array.

Examples:

add()
subtract()
multiply()
sqrt()
exp()
10. Statistical Functions

NumPy provides functions for statistical calculations:

mean() – average
median() – middle value
std() – standard deviation
var() – variance
sum() – total
min() – smallest value
max() – largest value
percentile() – percentage-based value
11. Moving Average

A moving average calculates the average of values within a fixed window that moves through the data.

It is useful for understanding trends in data.

12. Daily Returns

Daily return shows the percentage change between two consecutive values, commonly used in stock market data.

13. Volatility

Volatility shows how much the values change over time.

Usually, it is measured using the standard deviation of returns. Higher volatility means greater variation.

14. Random Module

numpy.random is used to generate random numbers and random arrays.

It is useful for simulation, testing, and sampling.
```

```python
Pandas Theory – Simple Language for Practical Exam
1. Pandas

Pandas is a Python library used for working with structured and tabular data. It is built on top of NumPy and makes data analysis easy.

2. Series

A Series is a one-dimensional labeled array. It can store different types of data and is similar to a single column in a table.

Simple: Series = One column of data.

3. DataFrame

A DataFrame is a two-dimensional table with rows and columns. It is similar to an Excel spreadsheet or SQL table.

Simple: DataFrame = Complete table with rows and columns.

4. Data Loading

Pandas provides functions to load data from different files:

read_csv() – reads CSV files
read_excel() – reads Excel files
read_json() – reads JSON files
5. Data Inspection

These functions help us understand the dataset:

head() – displays first 5 rows
tail() – displays last 5 rows
info() – gives information about columns and data types
describe() – gives statistical summary
shape – shows number of rows and columns
6. Indexing and Selection

Pandas provides different ways to select data:

loc[] – selects data using row and column labels
iloc[] – selects data using row and column positions
Boolean filtering – selects rows based on a condition

Example: Selecting students whose marks are greater than 80.

7. GroupBy

The groupby() function divides data into groups based on a particular column.

For example, we can group students according to their Gender and calculate their average marks.

8. Aggregation Functions

Aggregation functions are used to summarize data.

Common functions are:

sum() – calculates total
mean() – calculates average
count() – counts values
min() – finds minimum value
max() – finds maximum value
agg() – applies multiple functions together
9. Pivot Table

A pivot table is used to summarize and reorganize data.

It groups data and performs calculations, similar to an Excel Pivot Table.

10. Handling Missing Data

Sometimes datasets contain missing values. Pandas provides:

isnull() – checks missing values
dropna() – removes missing values
fillna() – fills missing values
11. Data Transformation

Data transformation means changing or improving the data.

Important functions are:

apply() – applies a function to data
map() – maps or changes values
astype() – changes the data type
rename() – changes column names
sort_values() – sorts the data
12. Merging and Joining

Pandas can combine multiple DataFrames.

merge() – combines DataFrames using a common column
join() – joins DataFrames based on indexes or keys
concat() – combines DataFrames row-wise or column-wise
```

```python
Central Tendency Theory – Simple Language for Practical Exam
1. Central Tendency

Central Tendency is a statistical method used to find one value that represents the whole dataset.

The three main measures are:

Mean
Median
Mode
2. Mean

Mean is the average of all values.

Formula:

Mean = Sum of all values / Total number of values

Example:
Values = 10, 20, 30

Mean = (10 + 20 + 30) / 3 = 20

👉 Mean uses all values, but it can be affected by very large or very small values called outliers.

3. Median

Median is the middle value after arranging the data in ascending or descending order.

For an odd number of values:

The middle value is the median.

Example:
10, 20, 30, 40, 50
Median = 30

For an even number of values:

Median is the average of the two middle values.

Example:
10, 20, 30, 40

Median = (20 + 30) / 2 = 25

👉 Median is not highly affected by outliers.

4. Mode

Mode is the value that occurs most frequently in a dataset.

Example:
10, 20, 20, 30, 40

Mode = 20

A dataset can have:

Unimodal – one mode
Bimodal – two modes
Multimodal – more than two modes

👉 Mode can also be used for categorical data, such as the most common color or product.

5. Choosing the Right Measure
Mean

Mean is best when the data is:
         
Symmetric
Without extreme values
Median

Median is best when:

Data has outliers
Data is skewed

Examples: Salary and house prices

Mode

Mode is best when we want to find:

Most common value
Most common category

Examples: Most sold product or most common grade

6. Effect of Outliers

An outlier is an extremely high or low value in a dataset.

Example:

10, 20, 30, 40, 1000

The value 1000 will greatly affect the mean.

But the median will not be affected much.

👉 Therefore, for data with outliers, median is usually better than mean.

7. Skewness and Relationship Between Mean, Median, and Mode
Symmetric Distribution

Mean = Median = Mode

All three values are approximately equal.

Right or Positive Skew

Mean > Median > Mode

The data has a longer tail on the right side.

Left or Negative Skew

Mean < Median < Mode

The data has a longer tail on the left side.

8. Grouped Data

For grouped or frequency data, data is divided into different classes.

Mean is calculated using class midpoints and frequencies.
Median is calculated using cumulative frequency and the median class.
Mode is calculated using the modal class.
9. Real-World Applications
Mean

Used for:

Average student marks
Average sales
Average performance
Median

Used for:

Salary analysis
House prices
Income analysis
Mode

Used for:

Most popular product
Most common grade
Most common category
```

```python
Measures of Variation Theory – Simple Language for Practical Exam
1. Measures of Variation (Dispersion)

Measures of Variation are statistical measures used to show how much the data values are spread out or scattered.

They help us understand whether the data is:

Consistent – values are close together
Variable – values are spread far apart

The main measures are:

Range
Variance
Standard Deviation
2. Range

Range is the difference between the maximum and minimum values in a dataset.

Formula:

Range = Maximum Value – Minimum Value

Example:

Data: 10, 20, 30, 40, 50

Range = 50 – 10 = 40

👉 Range is easy to calculate, but it can be affected by outliers.

3. Variance

Variance measures how far the data values are spread from the mean.

It calculates the average of the squared differences between each value and the mean.

Simple Meaning:
Small variance = data is more consistent
Large variance = data is more spread out

Variance uses squared values, so it is sometimes harder to understand directly.

4. Standard Deviation

Standard Deviation is the square root of variance.

It shows how much the data values usually differ from the mean.

Simple Meaning:
Small Standard Deviation → values are close to the mean → more consistent
Large Standard Deviation → values are far from the mean → less consistent

👉 Standard deviation is easier to understand than variance because it is expressed in the same unit as the original data.

5. Coefficient of Variation (CV)

The Coefficient of Variation (CV) compares the variability of different datasets.

Formula:

CV = (Standard Deviation / Mean) × 100

It is expressed as a percentage.

Simple Meaning:

A lower CV means the data is more consistent.

CV is useful when comparing datasets with different means or different units.

6. Interquartile Range (IQR)

The Interquartile Range (IQR) shows the spread of the middle 50% of the data.

Formula:

IQR = Q3 – Q1

Where:

Q1 = First Quartile
Q3 = Third Quartile

IQR is useful for finding and detecting outliers.

7. Population and Sample
Population

A population means the complete set of data.

For population variance and standard deviation, we use all the data values.

Sample

A sample is only a part of the complete population.

For sample variance and standard deviation, we generally use n – 1 in the calculation.

8. Importance of Measures of Variation

Measures of variation help us understand:

Data consistency
Risk
Reliability
Quality of data
Differences between datasets

They are important because the mean alone cannot tell us how much the data is spread.
```
